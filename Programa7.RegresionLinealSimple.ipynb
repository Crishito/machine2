```python
# Importamos las librerias
import pandas as pd
from sklearn.linear_model import LinearRegression
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Creamos el conjunto de datos
df = pd.DataFrame({
    "horas":[1,2,3,4,5],
    "nota":[2,3,5,4,6]
})
x= df[["horas"]]
y= df[["nota"]]
print (x)
```

```python
# Creamos nuestro modelo
modelo = LinearRegression()
modelo.fit(x, y)
```

```python
#Imprimimos el interceptor y el coef
print("Interceptor: ", modelo.intercept_)
print("Pendiente:", modelo.coef_[0])
```

```python
#Valores predichos
y_pred = modelo.predict(x)
```

```python
#REalizamos la nueva predicción 7
horas=float(input("Ingrese las horas de estudio"))
prediccion = modelo.predict([[horas]])

#REalizamos la nueva predicción
#prediccion = modelo.predict([[6]])

print("Predicción:" , prediccion)
```

Graficar con matplolib el modelo y verificar si el nuevo valor corresponde a la predicción en la recta.

```python
#Creamos los puntos de la recta
x_linea = np.linspace(1,6,100)
x_linea_df = pd.DataFrame({"horas": x_linea})
y_linea = modelo.predict(x_linea_df)
```

# Creamos la Figura

```python
plt.figure(figsize = (10,6))

#Dibujamos datos reales
plt.scatter(
    x["horas"],
    y,
    s=100,
    color="blue",
    label = "Datos reales"
)

#Dibujamos Los valores predichos
plt.scatter(
    x["horas"],
    y_pred,
    s=100,
    color= "orange",
    label = "Valores predichos"
)


#REcta de regresión
plt.plot(
    x_linea,
    y_linea,
    color="red",
    linewidth=2,
)

#Prediccion nueva
plt.scatter( 
    horas,
    prediccion[0],
    s=250,
    color="green",
    marker="o",
)
```
